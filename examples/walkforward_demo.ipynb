# Walk-forward Demo

This notebook runs a small offline example. Numbers are illustrative only.

```python
from pinewf.data import load_ohlcv
from pinewf.engine import StrategyConfig, run_backtest
from pinewf.metrics import compute_metrics
from pinewf.report import render_html_report
from pinewf.walk_forward import walk_forward_optimize

df = load_ohlcv("examples/sample_btc_4h.csv")
cfg = StrategyConfig()
result = run_backtest(df, cfg)
metrics = compute_metrics(result.equity, result.trades, cfg.initial, len(df), "strategy")
wf = walk_forward_optimize(df, {"fast": [9, 20], "slow": [50, 100], "sl": [6, 8]}, cfg)
render_html_report(result, metrics, wf, out_path="walkforward_demo.html")
wf.head()
```

Use the report to compare OOS windows, degradation, and drawdown before trusting any parameter set. Pair a surviving strategy with Trawlkit for live alert workflows.
